## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** *Bagaimana tren rata-rata tingkat polusi PM2.5 secara bulanan selama periode observasi (tahun 2013-2017), dan pada bulan apa saja kualitas udara biasanya mencapai titik terburuk?*
  - **Specific:** Fokus secara spesifik pada metrik polutan utama (PM2.5) dan tren bulanannya.
  - **Measurable:** Jelas pengukurannya, yaitu menggunakan perhitungan nilai rata-rata (mean) dari PM2.5.
  - **Action-Oriented:** Mengetahui bulan dengan puncak polusi memungkinkan otoritas terkait untuk menyiapkan tindakan preventif, seperti kampanye wajib masker atau pengetatan aturan emisi industri di bulan-bulan tersebut.
  - **Relevant:** Sangat relevan untuk kebutuhan mitigasi kesehatan publik.
  - **Time-bound:** Memiliki batasan waktu yang jelas, yaitu sepanjang periode observasi dataset (2013-2017).

- **Pertanyaan 2:** *Bagaimana pengaruh kondisi cuaca, khususnya curah hujan (RAIN) dan kecepatan angin (WSPM), terhadap fluktuasi konsentrasi PM2.5 di udara?*
  - **Specific:** Menargetkan hubungan antara dua variabel cuaca spesifik (hujan dan angin) terhadap tingkat PM2.5.
  - **Measurable:** Dapat diukur dan dibuktikan menggunakan analisis korelasi statistik atau perbandingan distribusi data.
  - **Action-Oriented:** Jika terbukti ada korelasi kuat (misal: angin kencang menurunkan polusi), temuan ini bisa digunakan sebagai dasar pembuatan *dashboard* peringatan kualitas udara yang terintegrasi dengan ramalan cuaca.
  - **Relevant:** Memahami interaksi alam dan polusi sangat esensial untuk memprediksi kualitas udara secara dinamis.
  - **Time-bound:** Dievaluasi berdasarkan seluruh rentang waktu yang direkam di dalam dataset.

## Import Semua Packages/Library yang Digunakan

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
print("Library berhasil diimpor.")
```

## Data Wrangling

### Gathering Data

#### Load df ...

```python
url = "https://raw.githubusercontent.com/marceloreis/HTI/master/PRSA_Data_20130301-20170228/PRSA_Data_Aotizhongxin_20130301-20170228.csv"
df = pd.read_csv(url)

print("Informasi: Menampilkan 5 baris teratas dari dataset untuk inspeksi awal.\n")
df.head()
```

**Insight:** (Opsional)
- Dataset yang digunakan hanya dari 1 stasiun pemantau saja yaitu Aotizhongxin yang telah berhasil dimuat ke dalam dataframe.
- Berdasarkan sampel awal, dataset ini mencatat data secara per jam (terlihat dari adanya kolom `year`, `month`, `day`, dan `hour`).
- Terdapat kolom-kolom yang merekam variabel polutan (seperti `PM2.5`, `PM10`, `SO2`, dll) serta variabel cuaca (seperti `TEMP`, `PRES`, `RAIN`, dll).

### Assessing Data

#### Identifying missing values dan duplicate data problem

```python
print("Informasi dataset:")
df.info()

print("\n" + "="*40 + "\n")

# cek jumlah missing values di setiap kolom
print("Jumlah missing values per kolom:")
print(df.isnull().sum())

# cek jumlah data yang duplikat
print("\nJumlah data duplikat:")
print(df.duplicated().sum())

# cek ringkasan parameter statistik
print("\nRingkasan Statistik:")
print(df.describe())
```

**Steps to Take:**
- **Menangani Missing Values**: Terdapat cukup banyak nilai kosong (NaN) pada kolom polutan (seperti `PM2.5`, `PM10`, `SO2`, `NO2`, `CO`, `O3`) dan beberapa variabel cuaca (seperti `wd`, `WSPM`, dll). Saya akan menangani nilai kosong ini dengan metode imputasi (mengisi dengan nilai *mean* atau *median*, atau metode *forward/backward fill* yang cocok untuk data *time-series*) pada tahap Cleaning Data.
- **Menangani Duplicate Data:** Berdasarkan pengecekan, tidak ditemukan adanya data yang terduplikasi (jumlah duplikat = 0). Oleh karena itu, tidak perlu ada tindakan penghapusan baris (drop duplicates).

**Insight:** (Opsional)
- Dataset memiliki total 35.064 baris data dengan 18 kolom.
- **Missing Values:** Terdapat nilai kosong (NaN) pada beberapa kolom polutan dan cuaca, dengan rincian aktual sebagai berikut:
    - PM2.5: 925 baris
    - PM10: 718 baris
    - SO2: 935 baris
    - NO2: 1023 baris
    - CO: 1776 baris
    - O3: 1719 baris
    - TEMP, PRES, DEWP, RAIN: masing-masing 20 baris
    - wd (arah angin): 81 baris
    - WSPM (kecepatan angin): 14 baris
- **Duplicate Data:** Tidak ditemukan adanya baris data yang terduplikasi (0 duplikat).
- **Data Type:** Seluruh tipe data sudah sesuai dengan fungsinya.

### Cleaning Data

#### Fixing missing values dan duplicate data problem

```python
# fix missing values pada kolom numerik (polutan dan cuaca)
numeric_cols = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3', 'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']
for col in numeric_cols:
  df[col] = df[col].interpolate(method='linear', limit_direction='both')

# fix missing values pada kolom kategorikal (arah angin)
df['wd'] = df['wd'].ffill().bfill()

# cek values
print("Status missing values after cleaning:")
print(df.isnull().sum())

print("\n" + "="*40 + "\n")
print("Data Cleaning Selesai! DataFrame sudah bersih dari NaN.")
```

**Insight:** (Opsional)
- Proses pembersihan data berfokus pada penanganan *missing values* karena tidak ditemukan data yang terduplikasi.
- Nilai kosong pada kolom numerik (polutan dan cuaca) berhasil ditangani menggunakan metode **interpolasi linear**. Metode ini sangat ideal untuk data *time-series* karena mampu memperkirakan nilai yang hilang berdasarkan tren data sebelum dan sesudahnya secara proporsional.
- Nilai kosong pada kolom kategorikal (`wd` / arah angin) diisi menggunakan metode **Forward Fill** dan **Backward Fill**.
- Pengecekan akhir membuktikan bahwa seluruh kolom sudah bersih (0 *missing values*). DataFrame kini sudah solid dan siap untuk dieksplorasi lebih jauh.

## Exploratory Data Analysis (EDA)

### Explore Data Kualitas Udara (PM2.5) dan Cuaca

```python
print("Informasi: Memulai proses Exploratory Data Analysis (EDA).\n")

# tren PM2.5 secara bulanan
print("Rata-rata tingkat polusi PM2.5 per bulan:")
monthly_pm25 = df.groupby('month')['PM2.5'].mean().reset_index()
display(monthly_pm25)

print("\n" + "="*40 + "\n")

# Korelasi antara PM2.5, Curah Hujan (RAIN), dan Kecepatan Angin (WSPM)
print("Matriks korelasi antara PM2.5, RAIN, WSPM:")
weather_corr = df[['PM2.5', 'RAIN', 'WSPM']].corr()
display(weather_corr)
```

**Insight:** (Opsional)
- **Eksplorasi 1 (Tren PM2.5 Bulanan):** Rata-rata tingkat polusi PM2.5 mengalami fluktuasi setiap bulannya. Kualitas udara cenderung memburuk pada akhir tahun, mencapai puncaknya pada bulan Desember (rata-rata 102.70). Sebaliknya, kualitas udara paling bersih tercatat di bulan Agustus dengan rata-rata PM2.5 terendah (55.85).
- **Eksplorasi 2 (Korelasi Cuaca):** Matriks korelasi mengonfirmasi adanya hubungan negatif antara PM2.5 dengan faktor cuaca.
    - Korelasi dengan kecepatan angin (WSPM) berada di angka **-0.275**, menunjukkan bahwa hembusan angin memiliki peran yang lumayan dalam mengurai atau menurunkan konsentrasi polusi.
    - Namun, korelasi dengan curah hujan (RAIN) sangat lemah di angka **-0.013**. Hal ini mengindikasikan bahwa secara agregat, hujan tidak memberikan dampak penurunan PM2.5 yang signifikan pada dataset ini.

## Visualization & Explanatory Analysis

### Pertanyaan 1: Tren PM2.5 Bulanan

```python
print("Informasi: Menghasilkan visualisasi Line Chart untuk tren PM2.5 bulanan.\n")

plt.figure(figsize=(10, 6))

# linechart dengan seaborn
sns.lineplot(
    x='month',
    y='PM2.5',
    data=monthly_pm25,
    marker='o',
    linewidth=2.5,
    color='firebrick'
)

# judul dan label
plt.title('Rata-rata Tingkat Polusi PM2.5 per Bulan\n(Stasiun Aotizhongxin 2013-2017)', fontsize=14, fontweight='bold')
plt.xlabel('Bulan', fontsize=12)
plt.ylabel('Rata-rata Konsentrasi PM2.5', fontsize=12)
plt.xticks(range(1, 13))
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show
```

### Pertanyaan 2: Korelasi Cuaca dan PM2.5

```python
print("Informasi: Menghasilkan visualisasi Scatter Plot untuk korelasi cuaca dan PM2.5.\n")

# subplot untuk menjejerkan 2 grafik secara horizontal
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# plot 1: kecepatan angin (WSPM) vs PM2.5
sns.scatterplot(x='WSPM', y='PM2.5', data=df, alpha=0.2, color='teal', ax=ax[0])
ax[0].set_title('Korelasi Kecepatan Angin vs PM2.5', fontsize=13, fontweight='bold')
ax[0].set_xlabel('Kecepatan Angin (WSPM)', fontsize=11)
ax[0].set_ylabel('Konsentrasi PM2.5', fontsize=11)

# plot 2: curah hujan (RAIN) vs PM2.5
sns.scatterplot(x='RAIN', y='PM2.5', data=df, alpha=0.2, color='royalblue', ax=ax[1])
ax[1].set_title('Korelasi Curah Hujan vs PM2.5', fontsize=13, fontweight='bold')
ax[1].set_xlabel('Curah Hujan (RAIN)', fontsize=11)
ax[1].set_ylabel('Konsentrasi PM2.5', fontsize=11)

plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- **Visualisasi 1 (Tren Bulanan):** *Line chart* dengan jelas memperlihatkan bentuk kurva menyerupai huruf 'U'. Tingkat PM2.5 mulai merangkak naik secara signifikan semenjak bulan September, kemudian mencapai titik terburuk di bulan Desember (musim dingin), dan perlahan turun hingga menyentuh titik terendah (kualitas udara terbaik) pada bulan Agustus.
- **Visualisasi 2 (Korelasi Cuaca):** - Pada *scatter plot* **Korelasi Kecepatan Angin vs PM2.5**, terlihat pola di mana konsentrasi PM2.5 yang ekstrem (di atas angka 400) sebagian besar hanya terjadi ketika kecepatan angin sangat rendah (di bawah 4 m/s). Seiring bertambahnya kecepatan angin (ke arah kanan sumbu X), titik data PM2.5 semakin berkurang dan menurun, mengonfirmasi korelasi negatif yang kita temukan sebelumnya.
    - Pada grafik **Korelasi Curah Hujan vs PM2.5**, titik data sangat terpusat pada angka curah hujan 0. Pola yang padat di sebelah kiri ini membuktikan bahwa meskipun ada korelasi negatif yang lemah, polusi tingkat tinggi jarang sekali terjadi secara bersamaan dengan turunnya hujan. Namun, karena hujan tidak sering terjadi dalam skala besar di dataset ini, hembusan angin tetap menjadi faktor yang lebih dominan dalam menyapu polusi.
